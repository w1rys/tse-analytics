```python
from delta import configure_spark_with_delta_pip
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
builder = (
    SparkSession.builder
    .master("local[*]")
    .appName("TSE Analytics Delta - Votação")
    .config(
        "spark.sql.extensions",
        "io.delta.sql.DeltaSparkSessionExtension",
    )
    .config(
        "spark.sql.catalog.spark_catalog",
        "org.apache.spark.sql.delta.catalog.DeltaCatalog",
    )
)

spark = configure_spark_with_delta_pip(builder).getOrCreate()

spark.sparkContext.setLogLevel("WARN")

print("Sessão Spark iniciada com sucesso!")
print("Versão do Spark:", spark.version)
print(
    "Pacotes:",
    spark.sparkContext.getConf().get("spark.jars.packages"),
)
```

```python
anos = list(range(2000, 2025, 2))

for ano in anos:
    print(f"Lendo {ano}...")

    df_raw_ano = (
        spark.read
        .format("csv")
        .option("header", "true")
        .option("sep", ";")
        .option("encoding", "ISO-8859-1")
        .option("inferSchema", "false")
        .load(
            f"data/{ano}/"
            f"votacao_candidato_munzona_{ano}/"
            f"*_BRASIL.csv"
        )
    )

    (
        df_raw_ano.write
        .format("parquet")
        .mode("overwrite")
        .save(f"data/raw/votacao_candidatos/{ano}")
    )

    print(f"{ano} concluído.")
```

```python
df_raw_votacao = (
    spark.read
    .format("parquet")
    .load("data/raw/votacao_candidatos/*")
)
```

```python
df_bronze_votacao = (
    df_raw_votacao
    .withColumn(
        "ANO_ELEICAO",
        F.col("ANO_ELEICAO").cast("int"),
    )
    .withColumn(
        "CD_TIPO_ELEICAO",
        F.col("CD_TIPO_ELEICAO").cast("int"),
    )
    .withColumn(
        "NR_TURNO",
        F.col("NR_TURNO").cast("int"),
    )
    .withColumn(
        "CD_ELEICAO",
        F.col("CD_ELEICAO").cast("int"),
    )
    .withColumn(
        "CD_CARGO",
        F.col("CD_CARGO").cast("int"),
    )
    .withColumn(
        "CD_MUNICIPIO",
        F.col("CD_MUNICIPIO").cast("int"),
    )
    .withColumn(
        "NR_ZONA",
        F.col("NR_ZONA").cast("int"),
    )
    .withColumn(
        "NR_CANDIDATO",
        F.col("NR_CANDIDATO").cast("int"),
    )
    .withColumn(
        "SQ_CANDIDATO",
        F.col("SQ_CANDIDATO").cast("long"),
    )
    .withColumn(
        "QT_VOTOS_NOMINAIS",
        F.col("QT_VOTOS_NOMINAIS").cast("long"),
    )
)
```

```python
(
    df_bronze_votacao.write
    .format("delta")
    .mode("overwrite")
    .save("data/bronze/votacao_candidatos")
)
```

```python
df_delta_votacao = (
    spark.read
    .format("delta")
    .load("data/bronze/votacao_candidatos")
)
```
